```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Admin\Documents\linkedin\Superstore Sales Dataset.csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["Sales"].sum()    # total sales 
```

```python
df.groupby("Category")["Sales"].sum()  # sales by category 
```

```python
df.groupby("City")["Sales"].sum().sort_values(ascending = False).head()   # top 5 cities 
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"],dayfirst = True)  # convert date 
```

```python
df.head()
```

```python
df.groupby(df["Order Date"].dt.month)["Sales"].sum()  # monthly sales sum()
```

# matplotlib

```python
df.groupby("Category")["Sales"].sum().plot(kind="bar")
plt.title("Sales by Category")
plt.show()
```
